```python
import numpy as np
def find_iris_win(h, cen=1402.77):
    wmin = np.array([h[f'TWMIN{win}'] for win in range(1, h['NWIN']+1)])
    wmax = np.array([h[f'TWMAX{win}'] for win in range(1, h['NWIN']+1)])
    # pdb.set_trace()
    return (np.where((wmin < cen)*(wmax > cen))[0][0] + 1).item()

def get_wave(h):
    return (np.arange(h['NAXIS1'])-h['CRPIX1'])*h['CDELT1'] + h['CRVAL1']

def get_time_sec(iris_file):
    nwin = iris_file[0].header['NWIN']
    time_loc = iris_file[nwin+1].header['TIME']
    return iris_file[nwin+1].data[:, time_loc]
```

```python
import copy

def set_zero(var0):
    var = copy.copy(var0)
    for i, item in enumerate(var):
        if isinstance(item, list):
            res = set_zero(item)
        else: 
            var[i] = 0
    return var
```

```python
def find_consecutive(arr0):
    arr = np.array(arr0)
    arr = arr * 1
    d1_arr = arr[1:] - arr[:-1]
    init_arr = np.where(d1_arr == 1)[0]+1
    fin_arr = np.where(d1_arr == -1)[0]+1
    return init_arr, fin_arr

def find_spikes(data0, sigma=3):
    grad = data0[1:] - data0[:-1]
    std = np.std(data0)
    inc = grad > sigma*std
    init, _ = find_consecutive(inc)
    dec = grad < -sigma*std
    _, fin = find_consecutive(dec) # should be subtract 1, but for index slice, keep it. 
    init += 1
    data1 = np.full(len(data0), False)
    for init0 in init:
        fin_gt = np.where(fin > init0)[0]
        if len(fin_gt) == 0: continue
        if (min(fin_gt) - init0) > 3: continue
        fin0 = fin[min(fin_gt)]
        data1[init0:fin0] = True
    return data1
```

```python
def iris_prep_despike(data, sigmas=4.5, Niter=10, kernel=None, min_std=1.0, 
                      silent=True, goodmap=False, mode='bright'):
    from scipy.signal import convolve
    from datetime import datetime

    t_begin = datetime.now()
    if not silent:
        print(f'{t_begin} IRIS_PREP_DESPIKE started on array of {data.size} elements.')
    result = copy.deepcopy(data)
    # result = np.where(result > 0, result, 0)
    where_bad = ~np.isfinite(data)                      
    result = np.where(np.isfinite(result), result, 1e-6)
    isize = data.shape
    Ndim = len(isize)
    match Ndim:
        case 1: 
            kernel = np.array([1.]*11)
        case 2: 
            kernel = np.array([1.]*9**2).reshape(9, 9)
        case 3: 
            kernel = np.array([1.]*5**3).reshape(5, 5, 5)
        case 4: 
            kernel = np.array([1.]*3**4).reshape(3, 3, 3, 3)
        case _:
            print(Ndim)
            print('Data dimensionality is too great. Cannot construct kernel.')
            return None
    goodmap = result*0 + 1.0
    for i in range(Niter):
        neighborhood_mean = convolve(goodmap*result, kernel, mode='same') \
                            / convolve(goodmap, kernel, mode='same')
        match mode:
            case 'bright': 
                deviation = result - neighborhood_mean      # find bright spikes only
            case 'dark':   
                deviation = neighborhood_mean - result      # find dark spikes only
            case 'both':   
                deviation = abs(result - neighborhood_mean) # find both bright and dark spikes only
            case _:
                print(f'Called with undefined mode: {mode}')
                return None
        neighborhood_std = np.sqrt(abs(convolve(goodmap*(deviation**2), kernel, mode='same')) \
                                 / (convolve(goodmap, kernel, mode='same')+1e-10))
        neighborhood_std = np.where(neighborhood_std > min_std, neighborhood_std, min_std)
        bad = deviation > (sigmas * neighborhood_std)
        if np.sum(bad) == 0: break
        newly_bad = np.logical_and(goodmap, bad)
        if np.sum(newly_bad) == 0: break
        if not silent:
            print(f'Iteration {i:4d} found {np.sum(bad):12d} bad pixels, {np.sum(newly_bad):12d} of them new.')
        goodmap[bad*newly_bad] = 0.0
    
        
    # Construct kernel k2 of the form exp(-r)/(1+r^Ndim). This will be used to construct
    # very-near-local smoothed version of data. It is just like the neighborhood_mean above,
    # except that the kernel is heavily weighted toward the nearest (good) pixels.
    
    Nk2 = 5  # size of very-near-local smoothing kernel
    middle = (Nk2-1)/2
    
    if Ndim == 1:
        k2 = np.exp(-abs(np.arange(Nk2) - middle))
    elif (Ndim > 1) and (Ndim <= 4):
        dim_array = [np.arange(isize[i]) for i in range(Ndim)]
        mesh_dim_array = np.meshgrid(*[np.arange(i) for i in isize], indexing='ij')
        r = np.sqrt(sum(dum**2 for dum in mesh_dim_array))
        k2 = np.exp(-r)/(1+r**Ndim)
    else: 
        print('Data dimensionality is too great. Cannot construct k2.')
        return None
              
    # We now re-use the neighborhood_mean array to conserve memory!
    neighborhood_mean = convolve(goodmap*result, kernel, mode='same') \
                        / convolve(goodmap, kernel, mode='same')
    
    # Replace bad pixels
    bad = np.logical_not(goodmap)
    if np.sum(bad) != 0:
        result[bad] = neighborhood_mean[bad]
    
    # Restore the bad crap that was filtered out of the data when we started. Also put the
    # bad crap into the result.
    if np.sum(where_bad) != 0:
       data[where_bad] = bad_crap
       result[where_bad] = bad_crap
    
    t_end = datetime.now()
    if not silent: 
        print(f'{t_end} IRIS_PREP_DESPIKE finished, {(t_end-t_begin).total_seconds()} sec elapsed.')
    # pdb.set_trace()
    return result
```

```python
spikes_scene = [False]*len(sg_files)
fit_fail = [False]*len(sg_files)
for moss_no in range(len(sg_files)):
    wave = si_info[moss_no][1]
    dw = np.mean(wave[1:] - wave[:-1])
    peak_ind = np.where(si_info[moss_no][2] == sg_ind[moss_no, 1])[0][0]
    net_flux = si_info[moss_no][4][peak_ind]*dw
    amp = si_info[moss_no][5][peak_ind][1]
    sigma = si_info[moss_no][5][peak_ind][3]
    g_flux = amp*sigma*np.sqrt(2*np.pi)
    if net_flux/g_flux > 1.5:
        spikes_scene[moss_no] = True
    t0_ind = iris2plus_moss.time_arg.sel(time=0, moss_no=moss_no).values
    t0_ind0 = np.argwhere(si_info[moss_no][2] == t0_ind)[0][0]
    if np.isnan(si_info[moss_no][-1][t0_ind0]) == True:
        fit_fail[moss_no] = True
print(f'spike scene = {np.sum(spikes_scene)}, ({np.sum(spikes_scene)*1e2/len(sg_files):.1f}%)')
print(f'fit fail scene = {np.sum(fit_fail)}, ({np.sum(fit_fail)*1e2/len(sg_files):.1f}%)')
```
